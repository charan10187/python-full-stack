```python
# FUNCTION IS A REUSABLE CODE BlocK  EX: SPOON
class spoon:
    def __init__(self,NAME,AGE):
        self.NAME=NAME
        self.AGE=AGE
        print(NAME,'spoon',AGE ,'k')
```

```python
print(spoon('silver',18))
```

```python
spoon('gold',22)
```
