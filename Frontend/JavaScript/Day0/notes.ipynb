## Break It 2
let name = "Sricharan";

let name = "Rahul";

console.log(name);

You chose B — Rahul.

❌ This one is not correct.

The correct answer is:

C — Error

Why?

You created the variable name twice using let:

let name = "Sricharan";

let name = "Rahul";

JavaScript doesn't allow you to declare the same let variable twice in the same scope.

You'll see an error similar to:

Identifier 'name' has already been declared

Don't worry about memorizing the word scope yet. We'll learn it later.


## Apply

You wrote:

    let name = "My name is Sricharan";
    let age = "Iam 20 years old";
    let favoriteColor = "My favorite color is blue";
    let learning = "I am learning JavaScript";

    console.log(name);
    console.log(age);
    console.log(favoriteColor);
    console.log(learning);

✅ Your code works.

However, there's something I want you to notice.

You stored:

let age = "Iam 20 years old";

Because you put quotes around it, it's a string.

You're not actually storing the age 20.

A better approach is:

    let name = "Sricharan";
    let age = 20;
    let favoriteColor = "blue"; 
    let learning = "JavaScript";

Now the variables contain the actual pieces of information:

    name          → "Sricharan"
    age           → 20
    favoriteColor → "blue"
    learning      → "JavaScript"

This will become extremely useful when we start doing calculations and combining information.
