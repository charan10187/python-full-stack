**Q1 — What is JavaScript used for?**

You said:

JavaScript is used in frontend for building the logic and interaction.

✅ Correct.

I'd make your definition slightly more precise:

JavaScript is used to add logic, behavior, and interactivity to websites.

For example:

HTML       → Button exists
CSS        → Button looks nice
JavaScript → Button does something when clicked

One small point: JavaScript isn't limited to frontend. Later, 
you'll learn that JavaScript can also run on servers using technologies 
such as Node.js. For now, frontend JavaScript is our focus.

__Q6 — Find the problem__

You said:

the double quotes has not enclosed properly

✅ Exactly.

This is incorrect:

console.log("Hello);

We started the string with:

"

but never closed it.

Correct:

console.log("Hello");

Think of quotes like a pair:

    " Hello "
    ↑       ↑
    start   end
