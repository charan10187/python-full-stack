# History of JavaScript in an Interview Perspective

JavaScript was created by **Brendan Eich** in **1995** while he was working at Netscape. It was initially developed to add interactive behavior to web pages.

## Key Timeline

- **1995:** JavaScript was created by Brendan Eich for Netscape Navigator.
- **1997:** JavaScript was standardized as **ECMAScript (ES)** by ECMA International.
- **1999:** ECMAScript 3 introduced important features such as regular expressions and exceptions.
- **2009:** ECMAScript 5 added JSON support, strict mode, and array methods.
- **2015:** ECMAScript 2015—also known as **ES6**—introduced major features such as:
    - `let` and `const`
    - Arrow functions
    - Classes
    - Promises
    - Modules
    - Template literals
- **2015 onward:** New features have been released annually through ECMAScript updates.

## JavaScript Today

JavaScript is no longer limited to web browsers. It is also used for:

- Front-end development with React, Angular, and Vue
- Back-end development with Node.js
- Mobile applications
- Desktop applications
- Game development
- Serverless applications
